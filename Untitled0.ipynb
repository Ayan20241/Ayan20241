```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import tensorflow as tf
from imblearn.over_sampling import RandomOverSampler
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("/content/diabetes.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
for i in range(len(df.columns[:-1])):
  label = df.columns[i]
  plt.hist(df[df['Outcome']==1][label], color='blue', label="Diabetes", alpha=0.7, density=True, bins=30)
  plt.hist(df[df['Outcome']==0][label], color='red', label="No Diabetes", alpha=0.7, density=True, bins=30)
  plt.title(label)
  plt.ylabel("Probability")
  plt.xlabel(label)
  plt.legend()
  plt.show()
```

```python
x = df[df.columns[:-1]].values
y = df[df.columns[-1]].values
```

```python
scalar = StandardScaler()
x_scaled = scalar.fit_transform(x)
data = np.hstack((x, np.reshape(y,(-1, 1))))
transform_df = pd.DataFrame(data, columns=df.columns)
```

```python
transform_df.head()
```

```python
over = RandomOverSampler()
x, y = over.fit_resample(x, y )
data = np.hstack((x, np.reshape(y,(-1, 1))))
transform_df = pd.DataFrame(data, columns=df.columns)
```

```python
transform_df.head()
```

```python
x_train, x_temp, y_train, y_temp = train_test_split(x, y, test_size=0.4, random_state=0)
x_valid, x_test, y_valid, y_test = train_test_split(x_temp, y_temp, test_size=0.5, random_state=0)
```

```python
model = tf.keras.Sequential([tf.keras.layers.Dense(256, activation='relu'),
                             tf.keras.layers.Dense(128, activation='relu'),
                             tf.keras.layers.Dense(1, activation='sigmoid')
])
```

```python
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss=tf.keras.losses.BinaryCrossentropy(),
              metrics=['accuracy'])
```

```python
model.evaluate(x_train, y_train)
```

```python
model.evaluate(x_valid, y_valid)
```

```python
model.fit(x_train, y_train, epochs=100, batch_size=32, validation_data=(x_valid, y_valid))
```
